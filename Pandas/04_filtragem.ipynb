# 04 - Filtragem de Dados

## Objetivo
Selecionar subconjuntos de linhas com base em condicoes.

## Conceitos

### Filtragem booleana
Uma condicao aplicada a uma Series retorna uma Series booleana.
Passar essa Series entre colchetes ao DataFrame retorna as linhas
onde a condicao e verdadeira.

### Operadores logicos
- `&` (e), `|` (ou), `~` (nao).
- Cada condicao deve estar entre parenteses por causa da precedencia.

### Metodos uteis
- `isin(lista)`: pertence a um conjunto.
- `between(a, b)`: intervalo inclusivo.
- `str.contains`, `str.startswith`, `str.endswith`: filtros em texto.
- `query("expressao")`: filtro com sintaxe de string.
- `loc` com condicao: `df.loc[cond, ["col1", "col2"]]`.

### Ordem importa
Aplique primeiro os filtros mais restritivos para reduzir o volume
de dados processados nas etapas seguintes.

### Combinando filtros
- `df[(cond1) & (cond2)]`.
- `df[cond1 & ~cond2]`.
- `df.query("coluna > 10 and outra == 'X'")`.

## DataFrame de exemplo
Criamos um DataFrame com nome, idade, salario, cidade e departamento
para praticar os filtros a seguir.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carla", "Diego", "Elisa", "Fabio"],
    "idade": [22, 25, 23, 28, 21, 35],
    "salario": [3500, 4200, 3900, 5100, 3300, 7200],
    "cidade": ["SP", "RJ", "MG", "SP", "RJ", "MG"],
    "departamento": ["TI", "RH", "TI", "Vendas", "RH", "TI"],
})
print("DataFrame:\n", df)
```

## Filtros simples
Um filtro booleano aplica uma condicao a uma coluna e retorna apenas
as linhas onde a condicao e `True`. Funciona tanto para numeros
quanto para texto.

```python
# Filtro simples
maiores_25 = df[df["idade"] > 25]
print("\nIdade > 25:\n", maiores_25)

# Filtro em texto
de_sp = df[df["cidade"] == "SP"]
print("\nCidade SP:\n", de_sp)
```

## Combinando condicoes
Usamos `&` (e), `|` (ou) e `~` (nao). Cada condicao precisa estar
**entre parenteses** por causa da precedencia dos operadores.

```python
# Multiplas condicoes
filtro = df[(df["idade"] > 22) & (df["salario"] > 4000)]
print("\nIdade > 22 e salario > 4000:\n", filtro)

# Ou logico
filtro_or = df[(df["cidade"] == "SP") | (df["cidade"] == "MG")]
print("\nSP ou MG:\n", filtro_or)

# Negacao
nao_ti = df[~(df["departamento"] == "TI")]
print("\nNao TI:\n", nao_ti)
```

## isin e between
- `isin(lista)`: retorna `True` se o valor pertence ao conjunto.
- `between(a, b)`: intervalo **inclusivo** nos dois extremos.

```python
# isin
cidades_alvo = ["SP", "MG"]
print("\nisin:\n", df[df["cidade"].isin(cidades_alvo)])

# between
print("\nSalario entre 3500 e 5000:\n",
      df[df["salario"].between(3500, 5000)])
```

## Filtros em strings
O acessador `.str` permite aplicar funcoes de texto sobre toda uma coluna:
- `str.contains("a", case=False)`: contem a substring, ignorando maiusculas.
- `str.startswith("A")`: comeca com o prefixo.

```python
# Filtros em strings
print("\nNomes com 'a':\n",
      df[df["nome"].str.contains("a", case=False)])

print("\nNomes que comecam com 'A':\n",
      df[df["nome"].str.startswith("A")])
```

## query
`query("expressao")` permite escrever filtros como strings, usando
`and`, `or`, `not` em vez de `&`, `|`, `~`. O codigo fica mais legivel
em condicoes complexas.

```python
# query
print("\nquery idade > 22:\n", df.query("idade > 22"))
print("\nquery combinada:\n",
      df.query("idade > 22 and cidade == 'SP'"))
```

## loc com condicao e colunas especificas
Combinando `loc` com uma mascara booleana, selecionamos linhas que
satisfazem a condicao **e** apenas as colunas de interesse.

```python
# loc com condicao e colunas especificas
print("\nloc filtrado:\n",
      df.loc[df["salario"] > 4000, ["nome", "salario"]])
```

## Resetando o indice apos o filtro
Apos filtrar, o indice mantem os rotulos originais (com buracos).
`reset_index(drop=True)` recria um indice sequencial de 0 a N-1.

```python
# Resetando indice apos filtro
filtro_reset = df[df["salario"] > 4000].reset_index(drop=True)
print("\nFiltro com indice resetado:\n", filtro_reset)
```

## Filtro com valores nulos
`isna()` retorna True onde ha NaN; `notna()` retorna True onde **nao** ha.
Ambos podem ser usados como mascara booleana.

```python
# Filtro com valores nulos
df_nan = df.copy()
df_nan.loc[2, "salario"] = np.nan
print("\nCom NaN:\n", df_nan)
print("Linhas sem salario:\n",
      df_nan[df_nan["salario"].isna()])
print("Linhas com salario:\n",
      df_nan[df_nan["salario"].notna()])
```
