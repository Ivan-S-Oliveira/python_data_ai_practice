# 01 - Series no Pandas

## Objetivo
Entender a estrutura Series, que e a base de uma coluna do DataFrame.

## Conceitos

### O que e uma Series
Uma Series e um array unidimensional rotulado. Ela combina:
- Um array de valores (dados).
- Um array de indices (rotulos).
- Um tipo de dado (dtype).

E o equivalente a uma coluna de uma tabela.

## Criacao de Series
Series podem ser criadas a partir de listas, dicionarios ou arrays NumPy.
Tambem podemos definir um indice explicito e um nome para a Series.

```python
import pandas as pd
import numpy as np

# Criando Series a partir de lista
s1 = pd.Series([10, 20, 30, 40])
print("Serie padrao:\n", s1)

# Com indice explicito
s2 = pd.Series([10, 20, 30], index=["a", "b", "c"])
print("\nSerie com indice:\n", s2)

# A partir de dicionario
s3 = pd.Series({"matematica": 8, "python": 9, "estatistica": 7})
print("\nSerie de dicionario:\n", s3)

# A partir de array NumPy
s4 = pd.Series(np.array([1.5, 2.5, 3.5]), name="valores")
print("\nSerie com nome:\n", s4)
```

## Propriedades da Series
Acessamos os atributos fundamentais: `values`, `index`, `dtype`, `name` e `shape`.

```python
# Propriedades
print("\nvalues:", s3.values)
print("index:", list(s3.index))
print("dtype:", s3.dtype)
print("name:", s3.name)
print("shape:", s3.shape)
```

## Indexacao e fatiamento
- **Por rotulo**: `s["chave"]`.
- **Por posicao**: `s.iloc[0]`.
- **Fatiamento**: `s["a":"c"]` — em Series com indice de texto, o rotulo final **e incluido**.

```python
# Indexacao por rotulo
print("\nNota de python:", s3["python"])

# Indexacao por posicao
print("Primeiro elemento:", s3.iloc[0])

# Fatiamento
print("\nFatia 'matematica':'python':\n", s3["matematica":"python"])
```

## Filtragem booleana
Assim como no NumPy, podemos filtrar Series com condicoes booleanas.
A mascara retorna uma Series de `True`/`False`.

```python
# Filtragem booleana
notas = pd.Series([5, 7, 9, 4, 8, 6], name="notas")
print("\nNotas >= 7:\n", notas[notas >= 7])
```

## Operacoes vetorizadas
Operacoes aritmeticas sao aplicadas elemento a elemento, sem loops.
Escalares sao propagados (broadcasting).

```python
# Operacoes vetorizadas
print("\nNotas + 1:\n", notas + 1)
print("Notas * 2:\n", notas * 2)
```

## Alinhamento por indice
Diferente do NumPy, a soma de duas Series **alinha pelos rotulos**, nao pela posicao.
Rotulos que nao existem em ambas geram `NaN` no resultado.

```python
# Alinhamento por indice
a = pd.Series([1, 2, 3], index=["x", "y", "z"])
b = pd.Series([10, 20, 30], index=["y", "z", "w"])
print("\nSoma alinhada por indice:\n", a + b)
```

## Valores ausentes (NaN)
Pandas representa ausentes com `NaN`. As funcoes `isna`, `notna`,
`fillna` e `dropna` ajudam a inspecionar, preencher e remover esses valores.
Reducoes como `sum` e `mean` **ignoram** NaN por padrao.

```python
# Valores ausentes
s5 = pd.Series([1.0, np.nan, 3.0, np.nan, 5.0])
print("\nSerie com NaN:\n", s5)
print("isna:\n", s5.isna())
print("notna:\n", s5.notna())
print("Soma ignorando NaN:", s5.sum())
print("Media ignorando NaN:", s5.mean())
```

## Preenchendo e removendo NaN
- `fillna(valor)` substitui os NaN pelo valor dado.
- `dropna()` remove as linhas com NaN.

```python
# Preenchendo NaN
print("\nPreenchido com 0:\n", s5.fillna(0))
print("Preenchido com a media:\n", s5.fillna(s5.mean()))

# Removendo NaN
print("\nSem NaN:\n", s5.dropna())
```

## Metodos uteis
A Series oferece varios metodos para exploracao rapida:
- `unique()`: valores unicos.
- `value_counts()`: frequencia de cada valor.
- `sort_values()`: ordena pelos valores.
- `describe()`: estatisticas descritivas.

```python
# Metodos uteis
print("\nValores unicos:", notas.unique())
print("Contagem de valores:\n", notas.value_counts())
print("Ordenado:\n", notas.sort_values())
print("Describe:\n", notas.describe())
```
