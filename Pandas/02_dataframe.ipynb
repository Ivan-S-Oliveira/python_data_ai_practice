# 02 - DataFrame

## Objetivo
Dominar a estrutura DataFrame, que e uma tabela bidimensional rotulada.

## Conceitos

### O que e um DataFrame
Um DataFrame e uma colecao de Series que compartilham o mesmo indice.
Cada coluna tem um nome e um dtype proprio. E a estrutura central do Pandas,
equivalente a uma planilha ou tabela SQL.

### Criacao
- A partir de dicionarios de listas.
- A partir de listas de dicionarios.
- A partir de arrays NumPy com colunas e indices nomeados.
- A partir de arquivos (CSV, Excel, JSON, SQL).

## Criacao a partir de dicionario
A forma mais comum: um dicionario onde as **chaves** viram nomes de colunas
e os **valores** (listas) viram as linhas.

```python
import pandas as pd
import numpy as np

# Criando DataFrame a partir de dicionario
dados = {
    "aluno": ["Ana", "Bruno", "Carla", "Diego", "Elisa"],
    "matematica": [8, 6, 10, 7, 9],
    "python": [9, 8, 7, 6, 10],
    "estatistica": [7, 9, 8, 7, 9],
}
df = pd.DataFrame(dados)
print("DataFrame:\n", df)
```

## Inspecao inicial
Ferramentas para entender rapidamente a estrutura do DataFrame:
- `shape`, `columns`, `index`, `dtypes`.
- `head` / `tail` para ver as primeiras/ultimas linhas.
- `info` mostra tipos e contagem de nao nulos.
- `describe` traz estatisticas descritivas das colunas numericas.

```python
# Inspecao
print("\nShape:", df.shape)
print("Colunas:", list(df.columns))
print("Indice:", list(df.index))
print("Dtypes:\n", df.dtypes)

print("\nhead(3):\n", df.head(3))
print("\ntail(2):\n", df.tail(2))
print("\ninfo:")
df.info()
print("\ndescribe:\n", df.describe())
```

## Criacao a partir de lista de dicionarios
Cada dicionario vira uma **linha**. As chaves se tornam as colunas.

```python
# A partir de lista de dicionarios
registros = [
    {"nome": "Ana", "idade": 22},
    {"nome": "Bruno", "idade": 25},
    {"nome": "Carla", "idade": 23},
]
df2 = pd.DataFrame(registros)
print("\nDataFrame de lista de dicionarios:\n", df2)
```

## Selecao de colunas
`df["coluna"]` retorna uma **Series**. Para retornar um **DataFrame**
com um subconjunto de colunas, passamos uma lista de nomes: `df[["c1", "c2"]]`.

```python
# Selecao de coluna
print("\nColuna python:\n", df["python"])
print("Tipo:", type(df["python"]))

# Selecao de multiplas colunas
print("\nSubconjunto:\n", df[["aluno", "python"]])
```

## Selecao com loc e iloc
- **`loc`**: selecao por **rotulo** (nome do indice/coluna). O fatiamento com
  `loc` e **inclusivo** no rotulo final.
- **`iloc`**: selecao por **posicao inteira**. O fatiamento segue a regra
  do Python (fim exclusivo).

```python
# loc por rotulo
print("\nloc[0]:\n", df.loc[0])
print("loc[0, 'aluno']:", df.loc[0, "aluno"])
print("loc[0:2, ['aluno', 'python']]:\n",
      df.loc[0:2, ["aluno", "python"]])

# iloc por posicao
print("\niloc[1]:\n", df.iloc[1])
print("iloc[0:3, 0:2]:\n", df.iloc[0:3, 0:2])
```

## Criacao e remocao de colunas
- Para criar: `df["nova"] = valores` (aceita Series, listas ou expressoes vetorizadas).
- Para remover: `df.drop(columns=[...])` retorna um **novo** DataFrame.
- Para renomear: `df.rename(columns={...})`.

```python
# Criando nova coluna
df["media"] = (df["matematica"] + df["python"] + df["estatistica"]) / 3
print("\nCom coluna media:\n", df)

# Removendo coluna
df_sem_media = df.drop(columns=["media"])
print("\nSem coluna media:\n", df_sem_media.columns.tolist())

# Renomeando
df_renomeado = df.rename(columns={"matematica": "mat", "python": "py"})
print("Colunas renomeadas:", df_renomeado.columns.tolist())
```

## Ordenacao e amostragem
- `sort_values(coluna)`: ordena pelas linhas segundo uma coluna.
- `sample(n)`: retorna n linhas aleatorias (util com `random_state` para reprodutibilidade).

```python
# Ordenando por coluna
print("\nOrdenado por media:\n", df.sort_values("media", ascending=False))

# Amostra aleatoria
print("\nAmostra:\n", df.sample(2, random_state=42))
```

## Conversao de tipos
Muitas vezes colunas chegam como texto (ex.: de um CSV) e precisam ser
convertidas. `pd.to_numeric` e `pd.to_datetime` sao as funcoes mais usadas.

```python
# Convertendo tipos
df3 = pd.DataFrame({"valor": ["10", "20", "30"]})
print("\nAntes:\n", df3.dtypes)
df3["valor"] = pd.to_numeric(df3["valor"])
print("Depois:\n", df3.dtypes)
```
