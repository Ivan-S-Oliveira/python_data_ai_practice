# 09 - Ordenacao

## Objetivo
Ordenar DataFrames por uma ou mais colunas.

## Conceitos

### sort_values
Ordena por valores de colunas.
- `by`: coluna ou lista de colunas.
- `ascending`: True ou False, ou lista de booleanos.
- `inplace`: modifica no lugar (evite).
- `na_position`: "first" ou "last" para NaN.
- `kind`: "quicksort", "mergesort", "heapsort".
- `ignore_index`: reindexa o resultado.

### sort_index
Ordena pelo indice.
- `ascending`, `axis`, `level` para indices multi-nivel.

### Ordenacao multipla
`df.sort_values(by=["col1", "col2"], ascending=[True, False])`
permite direcoes diferentes por coluna.

### nlargest e nsmallest
Atalhos para "top n" e "bottom n" com base em uma coluna.

### rank
`df.rank()` atribui posicoes de ordem, util para criar rankings.

### Ordem personalizada
Para ordem categorica, use `pd.Categorical` com `categories` e `ordered`.

## DataFrame de exemplo
Criamos um DataFrame com nome, idade, salario e departamento para
praticar os diferentes tipos de ordenacao.

```python
import pandas as pd

df = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carla", "Diego", "Elisa", "Fabio"],
    "idade": [22, 35, 23, 28, 21, 35],
    "salario": [3500, 4200, 3900, 5100, 3300, 7200],
    "departamento": ["TI", "RH", "TI", "Vendas", "RH", "TI"],
})
print("Original:\n", df)
```

## Ordenacao simples
`sort_values("coluna")` ordena crescentemente por padrao.
`ascending=False` inverte para decrescente.

```python
# Ordenacao simples
print("\nPor idade:\n", df.sort_values("idade"))

# Descendente
print("\nPor salario (desc):\n",
      df.sort_values("salario", ascending=False))
```

## Ordenacao por multiplas colunas
Passando uma lista em `by`, a ordenacao e lexicografica: a segunda coluna
so e usada para desempatar a primeira. `ascending` tambem aceita lista,
permitindo direcoes diferentes por coluna.

```python
# Multiplas colunas
print("\nPor idade e salario:\n",
      df.sort_values(by=["idade", "salario"]))

# Direcoes diferentes
print("\nIdade asc, salario desc:\n",
      df.sort_values(by=["idade", "salario"],
                     ascending=[True, False]))
```

## ignore_index
Apos ordenar, o indice original e preservado (fora de ordem).
`ignore_index=True` recria um indice sequencial de 0 a N-1.

```python
# Reset do indice
print("\nCom ignore_index:\n",
      df.sort_values("idade", ignore_index=True))
```

## Tratando NaN na ordenacao
`na_position` controla onde os NaN aparecem:
- `"last"` (padrao): NaN vao para o final.
- `"first"`: NaN vao para o inicio.

```python
# NaN no final
df_nan = df.copy()
df_nan.loc[1, "salario"] = None
print("\nCom NaN:\n", df_nan)
print("NaN no final:\n",
      df_nan.sort_values("salario", na_position="last"))
print("NaN no inicio:\n",
      df_nan.sort_values("salario", na_position="first"))
```

## Ordenacao por indice
`sort_index()` ordena pelos rotulos do indice, nao pelos valores.
Aceita `ascending` e funciona tambem com MultiIndex (`level=`).

```python
# Ordenacao por indice
df_idx = df.set_index("nome")
print("\nOrdenado por indice:\n", df_idx.sort_index())
print("Indice desc:\n", df_idx.sort_index(ascending=False))
```

## nlargest e nsmallest
Atalhos para obter o **top n** e o **bottom n** de uma coluna, sem
precisar ordenar o DataFrame inteiro.

```python
# nlargest e nsmallest
print("\n3 maiores salarios:\n", df.nlargest(3, "salario"))
print("3 menores salarios:\n", df.nsmallest(3, "salario"))
```

## rank
`rank()` atribui posicoes de ordem a cada elemento. Com
`ascending=False`, o maior valor recebe rank 1. Empates recebem
a media das posicoes por padrao.

```python
# rank
df["rank_salario"] = df["salario"].rank(ascending=False)
print("\nCom rank:\n", df)
```

## Ordem personalizada com Categorical
Para uma ordem **nao alfabetica** (ex.: RH < TI < Vendas), convertemos
a coluna em `pd.Categorical` passando `categories` e `ordered=True`.

```python
# Ordem personalizada com Categorical
ordem = ["RH", "TI", "Vendas"]
df["departamento"] = pd.Categorical(df["departamento"],
                                    categories=ordem,
                                    ordered=True)
print("\nOrdenado por departamento customizado:\n",
      df.sort_values("departamento"))
```

## Estabilidade da ordenacao
`kind="mergesort"` garante ordenacao **estavel**: elementos com a mesma
chave mantem a ordem relativa original. Util quando se ordena em etapas.

```python
# Ordenacao com metodo estavel
print("\nMergesort (estavel):\n",
      df.sort_values("idade", kind="mergesort"))
```

## Top n por grupo
Combinando `sort_values` com `groupby(...).head(n)` obtemos as n
melhores linhas de cada grupo.

```python
# Top n por grupo
print("\nMaior salario por departamento:\n",
      df.sort_values("salario", ascending=False)
        .groupby("departamento", observed=True)
        .head(1))
```
