# 01 - Arrays no NumPy

## Objetivo
Entender o que é um array NumPy, como criá-lo e quais são suas propriedades básicas.

## Conceitos

### O que é o NumPy
NumPy (Numerical Python) é a biblioteca base para computação numérica em Python.
Ela fornece o objeto `ndarray`, que é um array multidimensional de elementos
do mesmo tipo. Praticamente todas as bibliotecas científicas do Python
(Pandas, Scikit-Learn, SciPy) são construídas sobre o NumPy.


### Por que não usar listas do Python
Listas são flexíveis, mas lentas para operações numéricas. Arrays NumPy:
- Armazenam dados de forma contígua na memória.
- Permitem operações vetorizadas (sem loops explícitos).
- São digitados (todos os elementos têm o mesmo tipo).
- Suportam múltiplas dimensões com facilidade.

### Formas de criar arrays
- A partir de listas ou tuplas: `np.array`.
- Arrays de zeros e uns: `np.zeros`, `np.ones`.
- Sequências: `np.arange`, `np.linspace`.
- Matriz identidade: `np.eye`.
- Array vazio (não inicializado): `np.empty`.

```python
import numpy as np

# Criando array a partir de uma lista
a = np.array([1, 2, 3, 4, 5])
print("Array:", a)
print("Tipo:", type(a))
```

### Propriedades importantes de um array
- `shape`: tupla com o tamanho de cada dimensão.
- `ndim`: número de dimensões (eixos).
- `dtype`: tipo dos elementos.
- `size`: número total de elementos.
- `itemsize`: bytes por elemento.

```python
# Propriedades basicas
print("shape:", a.shape)
print("ndim:", a.ndim)
print("dtype:", a.dtype)
print("size:", a.size)
print("itemsize:", a.itemsize)
```

### Arrays multidimensionais
Podemos criar arrays com mais de uma dimensão passando listas aninhadas
para `np.array`. O atributo `shape` indica o tamanho de cada eixo.

```python
# Array bidimensional
b = np.array([[1, 2, 3], [4, 5, 6]])
print("\nArray 2D:")
print(b)
print("shape:", b.shape)
print("ndim:", b.ndim)
```

### Arrays de zeros e uns
Muito úteis para inicializar estruturas. Recebem uma tupla com o formato desejado.

```python
# Arrays de zeros e uns
z = np.zeros((2, 3))
u = np.ones((3, 2))
print("\nZeros:\n", z)
print("Uns:\n", u)
```

### Sequências numéricas
- `np.arange(inicio, fim, passo)`: similar ao `range`, mas retorna um array. O `fim` é exclusivo.
- `np.linspace(inicio, fim, quantidade)`: gera valores igualmente espaçados, incluindo o `fim`.

```python
# Sequencias
r = np.arange(0, 10, 2)          # inicio, fim (exclusivo), passo
l = np.linspace(0, 1, 5)         # inicio, fim, quantidade
print("\narange:", r)
print("linspace:", l)
```

### Matriz identidade
`np.eye(n)` cria uma matriz quadrada `n x n` com 1s na diagonal principal e 0s no restante.

```python
# Matriz identidade
i = np.eye(3)
print("\nIdentidade:\n", i)
```

### Especificando o dtype
Por padrão, o NumPy infere o tipo a partir dos dados. Podemos forçar um tipo
específico usando o parâmetro `dtype`.

```python
# Especificando dtype
c = np.array([1, 2, 3], dtype=np.float64)
print("\nCom dtype float:", c, c.dtype)
```

### Convertendo o tipo de um array
O método `astype` retorna um novo array com o tipo convertido
(o array original não é modificado).

```python
# Convertendo tipo
d = c.astype(np.int32)
print("Convertido para int:", d, d.dtype)
```
