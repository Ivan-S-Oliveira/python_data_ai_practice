# 07 - Números Aleatórios

## Objetivo
Gerar números aleatórios e amostrar distribuições com NumPy.

## Conceitos

### Gerador moderno
O NumPy recomenda usar `np.random.default_rng()` para criar um gerador
com estado controlável. Isso substitui as funções antigas como
`np.random.seed` e `np.random.rand`.

### Reprodutibilidade
Passar uma semente (`seed`) garante que a sequência gerada seja sempre
a mesma. Isso é essencial para experimentos e testes.

```python
import numpy as np

# Gerador com semente para reprodutibilidade
rng = np.random.default_rng(seed=42)
```

### Principais funções
- `rng.random(shape)`: uniforme em [0, 1).
- `rng.integers(low, high, size)`: inteiros no intervalo.
- `rng.normal(loc, scale, size)`: distribuição normal.
- `rng.uniform(low, high, size)`: uniforme em [low, high).
- `rng.binomial(n, p, size)`: binomial.
- `rng.poisson(lam, size)`: Poisson.

```python
# Uniforme em [0, 1)
u = rng.random(5)
print("Uniforme [0, 1):", u)
```

```python
# Inteiros
i = rng.integers(low=1, high=11, size=10)
print("Inteiros 1..10:", i)
```

```python
# Normal
n = rng.normal(loc=0, scale=1, size=5)
print("Normal (0, 1):", n)
```

```python
# Uniforme em intervalo
uni = rng.uniform(low=-5, high=5, size=5)
print("Uniforme [-5, 5):", uni)
```

```python
# Binomial
b = rng.binomial(n=10, p=0.5, size=5)
print("Binomial (n=10, p=0.5):", b)
```

```python
# Poisson
p = rng.poisson(lam=3, size=5)
print("Poisson (lambda=3):", p)
```

### Distribuições em formato de matriz
As mesmas funções aceitam `size` como tupla para gerar arrays
multidimensionais, úteis para inicialização de pesos em redes neurais
e para simulações vetorizadas.

```python
# Matriz normal
m = rng.normal(loc=0, scale=1, size=(3, 3))
print("\nMatriz normal 3x3:\n", m)
```

### Estatísticas de uma amostra grande
Amostras grandes permitem verificar propriedades teóricas das
distribuições (por exemplo, média e desvio padrão da normal).

```python
# Estatísticas de uma amostra grande
amostra = rng.normal(loc=10, scale=2, size=10000)
print("\nMédia da amostra:", amostra.mean())
print("Desvio da amostra:", amostra.std())
```

### Amostragem
- `rng.choice(array, size, replace, p)`: amostra de um array.
- `rng.shuffle(array)`: embaralha no lugar.
- `rng.permutation(array)`: retorna cópia embaralhada.

```python
# Amostragem com e sem reposição
valores = np.array([10, 20, 30, 40, 50])
com_reposicao = rng.choice(valores, size=5, replace=True)
sem_reposicao = rng.choice(valores, size=3, replace=False)
print("\nCom reposição:", com_reposicao)
print("Sem reposição:", sem_reposicao)
```

```python
# Amostragem com probabilidades
probs = np.array([0.1, 0.1, 0.1, 0.1, 0.6])
pond = rng.choice(valores, size=10, p=probs)
print("Amostra ponderada:", pond)
```

```python
# Embaralhar
arr = np.arange(10)
rng.shuffle(arr)
print("\nEmbaralhado:", arr)
```

```python
# Permutação (retorna cópia)
perm = rng.permutation(10)
print("Permutação:", perm)
```
