# 06 - Matrizes e Álgebra Linear

## Objetivo
Realizar operações de álgebra linear com NumPy.

## Conceitos

### Multiplicação de matrizes
- `A @ B` ou `np.matmul(A, B)`: produto matricial.
- `A * B`: produto elemento a elemento (não é multiplicação matricial).

### Transposta
- `A.T`: transposta da matriz.


```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
```

```python
# Multiplicação elemento a elemento
print("A * B (elemento a elemento):\n", A * B)
```

```python
# Multiplicação matricial
print("\nA @ B (matricial):\n", A @ B)
print("np.matmul(A, B):\n", np.matmul(A, B))
```

```python
# Transposta
print("\nTransposta de A:\n", A.T)
```

### Determinante e traço
- `np.linalg.det(A)`: determinante.
- `np.trace(A)`: soma da diagonal principal.


```python
# Determinante e traço
print("\nDeterminante de A:", np.linalg.det(A))
print("Traço de A:", np.trace(A))
```

### Matriz identidade e inversa
- `np.eye(n)`: matriz identidade.
- `np.linalg.inv(A)`: inversa.


```python
# Identidade
I = np.eye(2)
print("\nIdentidade:\n", I)
print("A @ I:\n", A @ I)
```

```python
# Inversa
A_inv = np.linalg.inv(A)
print("\nInversa de A:\n", A_inv)
print("A @ A_inv (deve ser identidade):\n", np.round(A @ A_inv, 10))
```

### Resolução de sistemas lineares
`np.linalg.solve(A, b)` resolve `A x = b` de forma eficiente e estável.

```python
# Sistema linear A x = b
b = np.array([5, 11])
x = np.linalg.solve(A, b)
print("\nSolução de A x = b:", x)
print("Verificação A @ x:", A @ x)
```

```python
# Autovalores e autovetores
autovalores, autovetores = np.linalg.eig(A)
print("\nAutovalores:", autovalores)
print("Autovetores:\n", autovetores)
```

### Produto interno e normas
- `np.dot(a, b)`: produto escalar de vetores.
- `np.linalg.norm(v)`: norma euclidiana.

```python
# Produto interno de vetores
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
print("\nProduto interno:", np.dot(u, v))
print("Norma de u:", np.linalg.norm(u))
```

### Decomposições
- `np.linalg.eig`: autovalores e autovetores.
- `np.linalg.svd`: decomposição em valores singulares.
- `np.linalg.qr`: decomposição QR.

```python
# Decomposição SVD
U, S, Vt = np.linalg.svd(A)
print("\nSVD - U:\n", U)
print("S:", S)
print("Vt:\n", Vt)
```
