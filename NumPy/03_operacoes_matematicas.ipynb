# 03 - Operacoes Matematicas

## Objetivo
Compreender operacoes vetorizadas e funcoes universais (ufuncs) do NumPy.

## Conceitos

### Vetorizacao
Operacoes em NumPy sao aplicadas elemento a elemento sem necessidade de loops.
Isso e chamado de vetorizacao e e muito mais rapido que loops em Python.

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])
```

### Operadores aritmeticos
`+`, `-`, `*`, `/`, `//`, `%`, `**` funcionam elemento a elemento
quando aplicados a arrays do mesmo shape.

```python
# Operacoes elemento a elemento
print("Soma:", a + b)
print("Subtracao:", b - a)
print("Multiplicacao:", a * b)
print("Divisao:", b / a)
print("Divisao inteira:", b // a)
print("Resto:", b % a)
print("Potencia:", a ** 2)

# Escalar com array
print("\na * 10:", a * 10)
print("a + 100:", a + 100)
```

```python
# Escalar com array
print("\na * 10:", a * 10)
print("a + 100:", a + 100)
```

### Funcoes universais (ufuncs)
Funcoes que operam elemento a elemento:
- `np.sqrt`, `np.exp`, `np.log`, `np.sin`, `np.cos`
- `np.abs`, `np.sign`, `np.floor`, `np.ceil`, `np.round`

```python
# Funcoes universais
x = np.array([1.0, 4.0, 9.0, 16.0])
print("\nRaiz:", np.sqrt(x))
print("Exponencial:", np.exp(np.array([0, 1, 2])))
print("Log natural:", np.log(np.array([1, np.e, np.e**2])))
print("Seno:", np.sin(np.array([0, np.pi/2, np.pi])))
print("Arredondar:", np.round(np.array([1.4, 1.5, 2.6])))
print("Piso:", np.floor(np.array([1.9, -1.1])))
print("Teto:", np.ceil(np.array([1.1, -1.9])))
print("Valor absoluto:", np.abs(np.array([-3, -1, 2])))
```

### Reducoes
Funcoes que reduzem um array a um unico valor:
- `np.sum`, `np.prod`, `np.min`, `np.max`
- `np.mean`, `np.std`, `np.var`
- Podem operar por eixo com o parametro `axis`.

```python
# Reducoes
v = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print("\nSoma:", np.sum(v))
print("Produto:", np.prod(v))
print("Minimo:", np.min(v))
print("Maximo:", np.max(v))
print("Media:", np.mean(v))
print("Desvio padrao:", np.std(v))
print("Variancia:", np.var(v))
```

```python
# Reducao por eixo
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print("\nMatriz:\n", m)
print("Soma total:", np.sum(m))
print("Soma por coluna (axis=0):", np.sum(m, axis=0))
print("Soma por linha (axis=1):", np.sum(m, axis=1))
print("Media por coluna:", np.mean(m, axis=0))
print("Media por linha:", np.mean(m, axis=1))
```

### Comparacoes
`>`, `<`, `==`, `!=`, `>=`, `<=` retornam arrays booleanos.
Combinacoes: `np.logical_and`, `np.logical_or`, `np.logical_not`.

```python
# Comparacoes
c = np.array([1, 5, 3, 8, 2])
print("\nMaiores que 3:", c > 3)
print("Iguais a 3:", c == 3)
print("Combinacao (maior que 2 e menor que 8):",
      np.logical_and(c > 2, c < 8))
```
