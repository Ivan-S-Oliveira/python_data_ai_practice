# 07 - Customizacao

## Conceito
Alem de plotar, precisamos ajustar o grafico para comunicar bem a informacao:
cores, fontes, tamanhos, grades, anotacoes, estilos e exportacao.

## Elementos que costumamos customizar
- Cores, marcadores, estilos de linha
- Rotulos, titulos, legendas
- Limites e ticks dos eixos
- Grade, spines (bordas)
- Anotacoes (ax.text, ax.annotate)
- Estilo global (plt.style.use)
- Exportacao (fig.savefig)

## Estilos disponiveis
plt.style.available

```python
import numpy as np
import matplotlib.pyplot as plt

# Listando estilos disponiveis
print(plt.style.available[:10])
```

```python
x = np.linspace(0, 2 * np.pi, 200)

with plt.style.context("seaborn-v0_8-whitegrid"):
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(x, np.sin(x), linewidth=2, label="seno")
    ax.plot(x, np.cos(x), linewidth=2, linestyle="--", label="cosseno")

    ax.set_xlabel("x (rad)")
    ax.set_ylabel("valor")
    ax.set_title("Funcoes trigonometricas")
    ax.legend(loc="upper right", frameon=True)

plt.show()
```

## Anotacoes

ax.annotate permite destacar um ponto com seta e texto. ax.text coloca
um texto livre em coordenadas do eixo.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), color="tab:blue")

# Ponto de maximo em x = pi/2
xmax = np.pi / 2
ymax = 1.0

ax.scatter([xmax], [ymax], color="red", zorder=5)
ax.annotate(
    "maximo local",
    xy=(xmax, ymax),
    xytext=(xmax + 1, ymax - 0.3),
    arrowprops=dict(arrowstyle="->", color="black"),
    fontsize=11,
)

ax.set_title("Anotacao em um grafico de linha")
ax.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

## Exportacao

fig.savefig salva a figura em disco. Formatos comuns: png, pdf, svg.
Use dpi para controlar resolucao e bbox_inches="tight" para evitar cortes.

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, np.sin(x))
ax.set_title("Figura exportada")

# Descomente para salvar
# fig.savefig("grafico.png", dpi=150, bbox_inches="tight")
plt.show()
```
