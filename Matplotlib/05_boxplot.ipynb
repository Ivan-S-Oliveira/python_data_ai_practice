# 05 - Boxplot

## Conceito
O boxplot resume a distribuicao de uma variavel a partir de cinco numeros:
minimo, primeiro quartil (Q1), mediana (Q2), terceiro quartil (Q3) e maximo,
alem de destacar outliers.

## Como ler
- Caixa: intervalo interquartil (IQR = Q3 - Q1), onde estao 50% dos dados
- Linha dentro da caixa: mediana
- Cercas (whiskers): Q1 - 1.5*IQR e Q3 + 1.5*IQR
- Pontos fora das cercas: possiveis outliers

## Quando usar
- Comparar distribuicoes entre grupos
- Detectar outliers
- Resumo robusto (menos sensivel a extremos que media)

## Funcoes principais
- plt.boxplot(dados)
- plt.violinplot(dados): alternativa que mostra a densidade

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)
grupo_a = rng.normal(0, 1, 200)
grupo_b = rng.normal(1.5, 1.2, 200)
grupo_c = rng.normal(-0.5, 0.8, 200)

fig, ax = plt.subplots(figsize=(7, 5))
ax.boxplot([grupo_a, grupo_b, grupo_c], tick_labels=["A", "B", "C"])

ax.set_ylabel("Valor")
ax.set_title("Comparacao de distribuicoes entre grupos")
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Boxplot horizontal

Use vert=False para deitar as caixas. Util quando os rotulos sao longos.

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.boxplot(
    [grupo_a, grupo_b, grupo_c],
    tick_labels=["A", "B", "C"],
    orientation="horizontal",
    patch_artist=True
)

ax.set_xlabel("Valor")
ax.set_title("Boxplot horizontal")

plt.show()
```
