# 03 - Histograma

## Conceito
O histograma mostra a distribuicao de uma variavel numerica. Os dados sao
divididos em intervalos (bins) e a altura de cada barra indica a frequencia
de observacoes naquele intervalo.

## Quando usar
- Entender a forma da distribuicao (simetrica, assimetrica, bimodal)
- Detectar outliers e concentracoes
- Comparar distribuicoes

## Funcoes principais
- plt.hist(dados, bins=n): desenha o histograma
- bins: numero de intervalos ou lista de limites
- density=True: normaliza para densidade (area = 1)
- cumulative=True: histograma acumulado

```python
import numpy as np
import matplotlib.pyplot as plt

# Gerando dados de uma normal
rng = np.random.default_rng(42)
dados = rng.normal(loc=0, scale=1, size=1000)

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(dados, bins=30, color="tab:purple", edgecolor="black", alpha=0.8)

ax.set_xlabel("Valor")
ax.set_ylabel("Frequencia")
ax.set_title("Histograma de uma distribuicao normal")
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Comparando bins

O numero de bins altera a leitura do grafico:
- Poucos bins: perde detalhe
- Muitos bins: ruido excessivo

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, b in zip(axes, [5, 20, 80]):
    ax.hist(dados, bins=b, color="tab:orange", edgecolor="black")
    ax.set_title(f"bins = {b}")

plt.tight_layout()
plt.show()
```
