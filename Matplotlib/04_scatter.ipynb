# 04 - Grafico de Dispersao (Scatter)

## Conceito
O scatter plota pares (x, y) como pontos. Serve para investigar a relacao
entre duas variaveis numericas, identificar correlacao, agrupamentos e outliers.

## Quando usar
- Verificar correlacao entre variaveis
- Detectar clusters
- Analisar outliers
- Representar uma terceira dimensao via tamanho ou cor do ponto

## Funcoes principais
- plt.scatter(x, y, s=tamanho, c=cor, alpha=transparencia)
- c pode receber um vetor e ser colorido por colormap
- plt.colorbar(): barra lateral com a escala de cor

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
x = rng.normal(0, 1, 200)
y = 2 * x + rng.normal(0, 0.5, 200)  # correlacao positiva com ruido

fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(x, y, alpha=0.7, color="tab:blue", edgecolor="black")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Dispersao com correlacao positiva")
ax.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

## Codificando uma terceira variavel

A cor (parametro c) pode representar uma terceira variavel. Usamos cmap para
definir a paleta e colorbar para exibir a escala.

```python
tamanho = rng.uniform(20, 200, 200)
valor_z = rng.normal(0, 1, 200)

fig, ax = plt.subplots(figsize=(7, 5))
sc = ax.scatter(x, y, s=tamanho, c=valor_z, cmap="viridis", alpha=0.7)
fig.colorbar(sc, ax=ax, label="z")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Scatter com tamanho e cor variaveis")

plt.show()
```
