# 02 - Grafico de Barras (Bar Plot)

## Conceito
O grafico de barras compara valores entre categorias discretas. A altura
(ou comprimento) de cada barra representa a magnitude da categoria.

## Quando usar
- Comparar categorias (vendas por produto, populacao por cidade)
- Dados categoricos ou discretos
- Barras horizontais quando os rotulos sao longos

## Funcoes principais
- plt.bar(x, height): barras verticais
- plt.barh(y, width): barras horizontais
- plt.xticks(rotation=...): rotacionar rotulos

```python
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E"]
valores = [23, 45, 12, 67, 34]

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(categorias, valores, color="tab:green", edgecolor="black")

ax.set_xlabel("Categoria")
ax.set_ylabel("Valor")
ax.set_title("Valores por categoria")
ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

## Barras agrupadas

Para comparar duas series por categoria, deslocamos a posicao x de cada
conjunto de barras usando a largura da barra.

```python
import numpy as np
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D"]
serie1 = [20, 35, 30, 25]
serie2 = [25, 32, 34, 20]

# Posicoes no eixo x
x = np.arange(len(categorias))
largura = 0.35

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(x - largura / 2, serie1, largura, label="Serie 1")
ax.bar(x + largura / 2, serie2, largura, label="Serie 2")

ax.set_xticks(x)
ax.set_xticklabels(categorias)
ax.set_ylabel("Valor")
ax.set_title("Comparacao entre series")
ax.legend()

plt.show()
```
