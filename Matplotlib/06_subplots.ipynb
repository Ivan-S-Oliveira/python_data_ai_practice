# 06 - Subplots (Multiplos graficos)

## Conceito
Subplots permitem colocar varios graficos na mesma figura, organizados em
uma grade de linhas e colunas. Cada eixo (Axes) e independente.

## Padroes de uso
- plt.subplots(nrows, ncols): retorna fig e um array de axes
- fig, axes = plt.subplots(2, 2): axes e um array 2x2
- ax = axes[i, j]: acessa um subplot especifico
- plt.subplots(..., sharex=True, sharey=True): compartilha eixos

## Vantagens
- Comparacao direta entre graficos
- Economia de espaco
- Consistencia visual em relatorios

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)

fig, axes = plt.subplots(2, 2, figsize=(10, 6))

axes[0, 0].plot(x, np.sin(x), color="tab:blue")
axes[0, 0].set_title("seno")

axes[0, 1].plot(x, np.cos(x), color="tab:orange")
axes[0, 1].set_title("cosseno")

axes[1, 0].plot(x, np.tan(x), color="tab:green")
axes[1, 0].set_title("tangente")
axes[1, 0].set_ylim(-5, 5)

axes[1, 1].plot(x, np.sin(x) * np.cos(x), color="tab:red")
axes[1, 1].set_title("seno * cosseno")

plt.tight_layout()
plt.show()
```

## Eixos compartilhados

sharex=True faz com que todos os subplots compartilhem o mesmo eixo x,
facilitando a comparacao de series temporais.

```python
fig, axes = plt.subplots(3, 1, figsize=(8, 6), sharex=True)

axes[0].plot(x, np.sin(x))
axes[0].set_ylabel("seno")

axes[1].plot(x, np.cos(x))
axes[1].set_ylabel("cosseno")

axes[2].plot(x, np.sin(x) + np.cos(x))
axes[2].set_ylabel("soma")
axes[2].set_xlabel("x")

plt.tight_layout()
plt.show()
```
