# 01 - Grafico de Linha (Line Plot)

## Conceito
O grafico de linha e usado para mostrar a evolucao de uma variavel ao longo de
outra, geralmente o tempo. Cada ponto representa um par (x, y) e os pontos sao
conectados por segmentos de reta.

## Quando usar
- Series temporais (temperatura ao longo dos dias, preco de acoes, etc.)
- Tendencia de uma metrica continua
- Comparacao de multiplas series no mesmo eixo

## Funcoes principais
- plt.plot(x, y): desenha a linha
- plt.xlabel / plt.ylabel: rotulos dos eixos
- plt.title: titulo do grafico
- plt.legend: legenda das series
- plt.grid: grade de fundo

## Estrutura minima
1. Importar matplotlib.pyplot
2. Preparar os dados (listas, arrays ou Series)
3. Chamar plt.plot
4. Adicionar rotulos e titulo
5. Chamar plt.show

```python
# Importacao padrao do pyplot
import matplotlib.pyplot as plt
import numpy as np

# Dados: dias de uma semana e temperatura media
dias = [1, 2, 3, 4, 5, 6, 7]
temperatura = [22, 24, 23, 25, 27, 26, 24]

# Criando a figura e o eixo
fig, ax = plt.subplots(figsize=(8, 4))

# Plot da linha. marker coloca um simbolo em cada ponto.
ax.plot(dias, temperatura, marker="o", color="tab:blue", label="Temperatura")

# Rotulos e titulo
ax.set_xlabel("Dia")
ax.set_ylabel("Temperatura (C)")
ax.set_title("Temperatura media ao longo da semana")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

## Multiplas linhas

Basta chamar plot varias vezes antes de show. O matplotlib atribui cores
automaticamente na ordem do ciclo de cores.

```python
# Duas series no mesmo grafico
x = np.linspace(0, 2 * np.pi, 100)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), label="seno")
ax.plot(x, np.cos(x), label="cosseno", linestyle="--")

ax.set_xlabel("x (radianos)")
ax.set_ylabel("valor")
ax.set_title("Seno e cosseno")
ax.legend()
ax.grid(True)

plt.show()
```
